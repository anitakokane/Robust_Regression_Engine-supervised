```python
"""
1.what is Regularization in machine learning ? Why is it needed?
->Regularization is a technique used to prevent a machine learning
  model from overfitting. It adds a penalty to the model to keep it
  simple and improve its performance on new data.

 2.Difference between Ridge Regression (L2) and Lasso Regression (L1).
 ->Ridge (L2): It reduces the size of coefficients but does not make
   them exactly zero.
   Lasso (L1): It can make some coefficients zero, which helps in selecting
   important features.

3.What is Cross Validation and Why is it Important?
-> Cross-validation is a technique used to check how well a machine
   learning model performs on different parts of the data. It helps to
   detect overfitting and check model reliability.

4.explain the following Cross Validation techniques:

a).K-Fold Cross Validation
->In this method, data is divided into K equal parts. The model is
  trained on K-1 parts and tested on the remaining part.

b).Stratified K-Fold cross Validation :
->It divides data into K parts while maintaining the same class ratio
  in each part. It is useful for imbalanced datasets.

c). Leave-One-Out Cross- Validation
-> In this method, one data point is used for testing and all remaining
 data points are used for training. This process repeats for every data
 point.

d). Time Series Split
->It is used for time-based data. The model is trained on past data
  and tested on future data without changing the time order.

5.Why are tree-based models less sensitive to feature scaling?
->Tree-based models use conditions like greater than or less than to
  split data. They do not depend on the actual scale of features, so
  feature scaling is usually not required.

"""
```

## Part B

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("/content/Advanced_Regression_HousePrice_Dataset_3800 - Advanced_Regression_HousePrice_Dataset_3800.csv.csv")
df
```

```python
df.info()
```

```python
df["sale_date"]=pd.to_datetime(df["sale_date"])
```

```python
df.info()
```

```python
## Basic Preprocessing
```

```python
from sklearn.preprocessing import StandardScaler

scaler=StandardScaler()
for col in df.select_dtypes(include=np.number).columns:
  df[col]=scaler.fit_transform(df[[col]])
```

```python
df
```

```python
## X , Y split

x = df.drop(columns=["property_id", "sale_date", "house_price_inr"])  # numeric columns only
y = df["house_price_inr"]
```

```python
x ## Independent variables
```

```python
y  ## Dependent variables
```

```python
## Train test split

from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

## Part C

```python
## Ridge Regression

from sklearn.linear_model import Ridge

ridge=Ridge(alpha=0.5)
ridge.fit(x_train, y_train)
```

```python
y_pred_ridge=ridge.predict(x_test)
```

```python
## Lasso
from sklearn.linear_model import Lasso

lasso=Lasso(alpha=0.1)
lasso.fit(x_train,y_train)
```

```python
y_pred_lasso=lasso.predict(x_test)
```

```python
## Tune regularization parameter using cross validation

from sklearn.linear_model import RidgeCV, LassoCV

alphas = [0.001, 0.01, 0.1, 1, 10, 100, 1000]

ridge = RidgeCV(alphas=alphas, cv=5).fit(x_train, y_train)
lasso = LassoCV(alphas=alphas, cv=5, max_iter=10000).fit(x_train, y_train)

print("Best alpha Ridge:", ridge.alpha_)
print("Best alpha Lasso:", lasso.alpha_)
```

```python
from sklearn.metrics import root_mean_squared_error, r2_score

## Training
print("RMSE of Ridge:",root_mean_squared_error(y_train, ridge.predict(x_train)))
print("R2 Score of Ridge:", r2_score(y_train, ridge.predict(x_train)))

print("RMSE of Lasso:",root_mean_squared_error(y_train, lasso.predict(x_train)))
print("R2 Score of Lasso:", r2_score(y_train, lasso.predict(x_train)))
```

```python
## Validation
print("RMSE of Ridge:",root_mean_squared_error(y_test, y_pred_ridge))
print("R2 Score of Ridge:", r2_score(y_test, y_pred_ridge))

print("RMSE of Lasso:",root_mean_squared_error(y_test, y_pred_lasso))
print("R2 Score of Lasso:", r2_score(y_test, y_pred_lasso))
```

```python
## Feature Coefficient behavior

print("Coefficients of Ridge:")
for feature, coef in zip(x_train.columns, ridge.coef_):
    print(f"{feature}: {coef}")

print("\nCoefficients of Lasso:")
for feature, coef in zip(x_train.columns, lasso.coef_):
    print(f"{feature}: {coef}")
```

## Part D

```python
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import KFold
from sklearn.pipeline import make_pipeline

model = make_pipeline(StandardScaler(), Ridge(alpha=1))
kf = KFold(n_splits=5, shuffle=True, random_state=42)

scores_kfold = cross_val_score(model, x, y, cv=kf,scoring="neg_mean_squared_error")
print("rmse:",np.sqrt(-scores_kfold.mean()))
```

```python
## Stratified
from sklearn.model_selection import StratifiedKFold

y_bins = pd.qcut(y, q=5, labels=False)

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
scores_skf = cross_val_score(model, x, y, cv=skf.split(x, y_bins),scoring="neg_mean_squared_error")
print("rmse:",np.sqrt(-scores_skf.mean()))
```

```python
## Leave-One-Out

from sklearn.model_selection import LeaveOneOut

loo=LeaveOneOut()
scores_loo=cross_val_score(model,x,y,cv=loo,scoring="neg_mean_squared_error")
rmse_loo=np.sqrt(-scores_loo.mean())
print(rmse_loo)
```

```python
## Time series
from sklearn.model_selection import TimeSeriesSplit

df_time=pd.read_csv("/content/Advanced_Regression_HousePrice_Dataset_3800 - Advanced_Regression_HousePrice_Dataset_3800.csv.csv")

df_time=df_time.sort_values(by="sale_date")

x_time=df_time.drop(columns=["property_id","sale_date","house_price_inr"])
y_time=df_time["house_price_inr"]

tscv=TimeSeriesSplit(n_splits=5)
scores_tscv=cross_val_score(model,x_time,y_time,cv=tscv,scoring="neg_mean_squared_error")

print("rmse:",np.sqrt(-scores_tscv.mean()))
```

```python
## 14.Analyze how performance metrics vary different CV strategies

"""
Model performance was evaluated using RMSE across all CV techniques.
K-Fold, Stratified K-Fold and LOOCV gave similar RMSE values, showing
stable generalization. Stratified K-Fold had lower variance across folds
because each fold had a balanced price distribution. Time Series Split
gave a different RMSE because it preserves time order and uses less training
data in early folds.LOOCV was the most computationally expensive with little
gain in accuracy.”
"""
```

## Part E

```python
df_decision=pd.read_csv("/content/Advanced_Regression_HousePrice_Dataset_3800 - Advanced_Regression_HousePrice_Dataset_3800.csv.csv")

x_decision=df_decision.drop(columns=["property_id","sale_date","house_price_inr"])
y_decision=df_decision["house_price_inr"]

x_train_decision,x_test_decision,y_train_decision,y_test_decision=train_test_split(x_decision,y_decision,test_size=0.2,random_state=42)

from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_squared_error

decision_tree=DecisionTreeRegressor(max_depth=6)
decision_tree.fit(x_train_decision,y_train_decision)

y_pred_decision=decision_tree.predict(x_test_decision)
print("RMSE of Decision Tree:",root_mean_squared_error(y_test_decision, y_pred_decision))
print("MSE:",mean_squared_error(y_test_decision, y_pred_decision))
print("R2 Score of Decision Tree:", r2_score(y_test_decision, y_pred_decision))
```

```python
from sklearn.ensemble import RandomForestRegressor

random_forest=RandomForestRegressor(n_estimators=100,max_depth=30,random_state=42)
random_forest.fit(x_train_decision,y_train_decision)

y_pred_random_forest=random_forest.predict(x_test_decision)

print("RMSE of Random Forest:",root_mean_squared_error(y_test_decision, y_pred_random_forest))
print("MSE:",mean_squared_error(y_test_decision, y_pred_random_forest))
print("R2 Score of Random Forest:", r2_score(y_test_decision, y_pred_random_forest))
```

```python
## Single Tree vs ensemble performance

"""
The single Decision Tree (max_depth=6) gave an RMSE of about 28.6 lakh
and an R² of 0.899. The Random Forest ensemble gave a lower RMSE of about
24.2 lakh and a higher R² of 0.927 on the same test set. So the ensemble
predicts more accurately.

A single tree has high variance. A small change in the training data can
change its splits, and its depth of 6 limits it to coarse predictions.

Random Forest trains many trees on different random samples and features,
then averages their predictions. Averaging cancels out the individual
trees’ errors, which reduces variance and overfitting.
This lets the forest use deeper trees (depth 30) to capture complex
patterns without overfitting as badly as a single deep tree would.
"""
```

## Part F

```python
from sklearn.svm import SVR

linear=SVR(kernel="linear")
linear.fit(x_train,y_train)
linear_pred=linear.predict(x_test)

print("RMSE of Linear SVR:",root_mean_squared_error(y_test, linear_pred))
print("MSE:",mean_squared_error(y_test, linear_pred))
print("R2 Score of Linear SVR:", r2_score(y_test, linear_pred))
```

```python
rbf=SVR(kernel="rbf")
rbf.fit(x_train,y_train)
rbf_pred=rbf.predict(x_test)

print("RMSE of RBF SVR:",root_mean_squared_error(y_test, rbf_pred))
print("MSE:",mean_squared_error(y_test, rbf_pred))
print("R2 Score of RBF SVR:", r2_score(y_test, rbf_pred))
```

```python
## Compare SVR performance with linear and tree based models

"""
linear models(Ridge/Lasso):They are fast and easy to interpret, but they
can only fit straight-line relationships. If their RMSE is higher than the
others,it shows that house price depends on non-linear effects and
interactions, such as location and area together

Tree-based models: The Decision Tree captures non-linear patterns but is
unstable (high variance). Random Forest averages many trees, which reduced
RMSE from about 28.6 lakh to 24.2 lakh and raised R² from 0.899 to 0.927.
Neither needs feature scaling.

SVR: With the RBF kernel it captures non-linear patterns, so it should beat
the linear models.it can come close to the tree models. However, it needs
feature and target scaling,is sensitive to hyperparameters, and is slower on
large datasets.

"""
```

## Part G

```python
m, s = y_decision.mean(), y_decision.std(ddof=0)   # to convert scaled price back to rupees
```

```python
from sklearn.metrics import mean_absolute_error

m, s = y_decision.mean(), y_decision.std(ddof=0)

models = [
    ("Ridge", ridge, True),
    ("Lasso", lasso, True),
    ("Decision Tree", decision_tree, False),
    ("Random Forest", random_forest, False),
    ("Linear SVR", linear, True),
    ("RBF SVR", rbf, True),
]

results = []
for name, model, scaled in models:
    if scaled:
        y_pred = model.predict(x_test) * s + m       # back to rupees
    else:
        y_pred = model.predict(x_test_decision)      # already rupees
    y_true = y_test_decision                         # always rupees

    mse = mean_squared_error(y_true, y_pred)
    results.append({
        "Model": name,
        "MSE": mse,
        "MAE": mean_absolute_error(y_true, y_pred),
        "RMSE": np.sqrt(mse),
        "R2": r2_score(y_true, y_pred),
    })

df_result = pd.DataFrame(results).set_index("Model")
df_result.round(3)
```

```python
"""
Random Forest is the best on every metric (lowest MSE , MAE,and RMSE,
highest R2), and RBF SVR is a close second .The Decision Tree is the
Weakest.MSE and RMSE punish large errors more, MAE is the average error
in rupees(about 18 to 21 lakh) and r2 shows that all models explain
90 to 93% of the price variance.

"""
```

```python
## 23 Comparison

"""
Regularized linear (Ridge, Lasso): R² 0.919, RMSE about 25.6 lakh.They
are almost identical, and Lasso removed no feature (all coefficients
stayed non-zero), because the features are all useful and the best alpha
was small.

Tree-based: the single Decision Tree is the weakest (R² 0.899), but
Random Forest is the best overall (R² 0.927). Averaging many trees
reduced RMSE by about 4.4 lakh.

SVR: the linear SVR (0.917) matches the linear models, while RBF
SVR (0.925) is nearly as good as Random Forest because it captures
non-linear effects.
"""
```

```python
## Overfitting and Underfitting

rows = []
for name, model, scaled in models:
    if scaled:
        p_tr = model.predict(x_train) * s + m
        p_te = model.predict(x_test) * s + m
    else:
        p_tr = model.predict(x_train_decision)
        p_te = model.predict(x_test_decision)
    rows.append([name, r2_score(y_train_decision, p_tr), r2_score(y_test_decision, p_te)])

pd.DataFrame(rows, columns=["Model", "Train R2", "Test R2"]).set_index("Model").round(3)
```

```python
"""
Ridge, Lasso, Linear SVR: train R² is about 0.916 and test R² about 0.919,
so the two are nearly equal. This is a good fit, with no overfitting.
Their R² is not very high, but since the RBF and Random Forest models only
reach 0.925 to 0.927,it is not clear underfitting either, only a small limit of
straight-line models.

Decision Tree (depth 6): train and test R² should be close, with the train
slightly higher. Depth 6 keeps it simple, so it leans towards mild underfitting
(coarse predictions), which matches its lowest test R².

Random Forest (depth 30): train R² will be much higher (about 0.98 to 0.99)
han test (0.927). That gap is a sign of overfitting, though the test score is
still the best because averaging controls it.

RBF SVR: train and test R² should be close, so it is well balanced.
"""
```
